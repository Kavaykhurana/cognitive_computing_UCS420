Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.

```python
dt = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15", "2017-07-19", "2013-06-01",
        "2021-02-10", "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df_employees = pd.DataFrame(dt)
df_employees.to_csv("employees.csv", index=False)

print("employees.csv created successfully!\n")
print(df_employees)
```

a) Shape (number of rows and columns) of the DataFrame.

```python
print("Shape of DataFrame (rows, columns):", df_employees.shape)
```

b) Summary of the DataFrame that includes the data types and non-null counts for each column.

```python
df_employees.info()
```

c) Generate descriptive statistics.

```python
print(df_employees.describe())
```

d) Display the first 5 rows and last 3 rows of the dataset.

```python
print("First 5 rows:")
print(df_employees.head(5))
print("\nLast 3 rows:")
print(df_employees.tail(3))
```

e) Calculate the following statistics from the dataset:
i. The average salary of employees.
ii. The total bonus paid to all employees.
iii. The youngest employee's age.
iv. The highest performance rating.

```python
print("Average Salary:", df_employees['Salary'].mean())
print("Total Bonus:", df_employees['Bonus'].sum())
print("Youngest Employee Age:", df_employees['Age'].min())
print("Highest Performance Rating:", df_employees['Rating'].max())
```

f) Sort the DataFrame by the Salary column in descending order.

```python
sorted_df = df_employees.sort_values(by='Salary', ascending=False)
print(sorted_df)
```

g) Add a new column that categorizes employees based on their performance rating:
i. Excellent for ratings >= 4.5
ii. Good for ratings >= 4.0 but < 4.5
iii. Average for ratings < 4.0

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df_employees['Performance_Category'] = df_employees['Rating'].apply(categorize_rating)
print(df_employees)
```

h) Identify missing values in the DataFrame.

```python
print("Missing values in each column:")
print(df_employees.isnull().sum())
```

i) Rename the Employee_ID column to ID.

```python
df_employees.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print(df_employees)
```

j) Find all employees who:
i. Have more than 5 years of experience.
ii. Belong to the IT department.

```python
filtered_df = df_employees[(df_employees['Years_of_Experience'] > 5) & (df_employees['Department'] == 'IT')]
print(filtered_df)
```

k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

```python
df_employees['Tax'] = df_employees['Salary'] * 0.10
print(df_employees)
```

l) Save the modified DataFrame (with added columns) to a new CSV file.

```python
df_employees.to_csv("modified_employees.csv", index=False)
print("Modified DataFrame saved to modified_employees.csv successfully!")
```
